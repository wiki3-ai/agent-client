# notebook-agent quickstart

This is the **user UX** for `notebook-agent`. Everything happens inside a
regular Python kernel — there is no CLI. You import the package, call
`run_task(...)`, and use the `show_*` helpers to inspect the result
right here in the notebook.

The agent follows the **Retrieve → Compose → Transform → Generate**
policy and persists every task as a filesystem-backed graph under
`runs/`.

```python
# Papermill parameters — override these when executing this notebook headless.
request = "Use the echo skill to echo hello from a Jupyter notebook"
message = "hello from a Jupyter notebook"
runs_root = "runs"
```

## 1. Run a task

`run_task` is the only entry point you need. It returns an `AgentResult`
carrying the task object, the chosen stage, and the structured outputs.

```python
from notebook_agent import run_task

# Headless-test hook: configure a DSPy DummyLM when NOTEBOOK_AGENT_DSPY_ANSWERS_JSON is set.
# In interactive use, leave DSPy to env vars or your own dspy.configure() call.
import os, json as _json
_answers = os.environ.get("NOTEBOOK_AGENT_DSPY_ANSWERS_JSON")
if _answers:
    import dspy
    from dspy.utils.dummies import DummyLM
    dspy.configure(lm=DummyLM(_json.loads(_answers)))

result = run_task(
    request,
    parameters={"message": message},
    runs_root=runs_root,
)
assert result.success, f"task failed: {result.error}"
result.task.directory
```

## 2. Inspect the task in the notebook

The `show_*` helpers return IPython display objects, so they render
natively in any Jupyter UI.

```python
from notebook_agent import show_task, show_answer, show_manifest, show_events, show_graph

show_task(result)
```

```python
show_answer(result)
```

```python
show_graph(result)
```

```python
show_events(result, tail=10)
```

```python
show_manifest(result)
```

## 3. Inspect the skill catalog

Selecting a skill is a DSPy step inside `run_task`; here we just list the
catalog the agent sees.

```python
from notebook_agent import SkillRepository
from notebook_agent.transform import builtin_skills_root

for skill in SkillRepository([builtin_skills_root()]).all_skills():
    print(f"{skill.skill_id:24s}  {skill.name}")
```

## 4. Re-open an existing run

Every run is just a directory under `runs/YYYY/MM/DD/`. Load it back with
`TaskGraph.load` and inspect it the same way.

```python
from notebook_agent import TaskGraph

graph = TaskGraph.load(result.task.directory)
show_graph(graph)
```
