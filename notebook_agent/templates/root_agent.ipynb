# notebook-agent root task template

This is the **Papermill root template** for the notebook-centric coding agent.
Executing this notebook IS the agent run: it builds a plan, searches local
skills, generates code when no skill matches, executes the resulting notebook,
and reports the answer.

State for this run is captured in the executed notebook's metadata under
`notebook_agent.*` and in the visible cells below.


```python
# Papermill parameters
prompt = "say hello from the root template"
runs_root = "runs"
max_autonomous_turns = 6
provider = None     # falls back to NOTEBOOK_AGENT_PROVIDER / lm_studio
model = None        # falls back to NOTEBOOK_AGENT_MODEL
base_url = None     # falls back to NOTEBOOK_AGENT_BASE_URL
api_key = None      # falls back to NOTEBOOK_AGENT_API_KEY
max_tokens = None   # falls back to NOTEBOOK_AGENT_MAX_TOKENS (16384)
temperature = None  # falls back to NOTEBOOK_AGENT_TEMPERATURE (0.0)
continue_from = None  # path to a prior task directory to continue
```

```python
from notebook_agent import run_task
from notebook_agent.litellm_client import LiteLLMClient

# Build LLM client only if the user wants to override env defaults.
_llm = None
_overrides = {
    k: v for k, v in dict(
        provider=provider, model=model, base_url=base_url, api_key=api_key,
        max_tokens=max_tokens, temperature=temperature,
    ).items() if v is not None
}
if _overrides:
    _llm = LiteLLMClient(**_overrides)

result = run_task(
    prompt,
    runs_root=runs_root,
    max_autonomous_turns=max_autonomous_turns,
    llm=_llm,
    continue_from=continue_from,
)

print(f"status: {'success' if result.success else 'failed'}")
print(f"stage : {result.stage_used}")
print(f"turns : {result.turns_used}")
print(f"task  : {result.task.directory}")
print()
print(result.answer or '(no answer)')
```
